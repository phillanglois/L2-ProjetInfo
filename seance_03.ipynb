## Rappel — Rééchantillonnage et interpolation

On a vu l'intérêt de la transformation inverse : pour chaque pixel de sortie, on cherche où lire dans l'image source. 
Cela évite "les trous" d'une transformation directe. 

Après une transformation ou un redimensionnement, une position de sortie correspond souvent à une position non entière dans l’image d’entrée. 
Comme les intensités ne sont connues qu’aux pixels de la grille, _ie_ aux coordonnées entières, l’interpolation estime la valeur cherchée à partir de valeurs connues :

- le plus proche voisin choisit le pixel le plus proche ;
- l'interpolation bilinéaire combine les quatre pixels voisins selon leurs distances.

### Plus proche voisin

$I'(x,y)=I(\operatorname{arrondi}(x),\operatorname{arrondi}(y)).$

C’est rapide et cela conserve les valeurs originales, mais peut produire des escaliers visibles (crénelage).

**Indications.** Comprendre l'_aliasing_ (crénelage) sur un schéma.

### Interpolation bilinéaire

Pour un point $(x, y)$ situé entre dans quatre sommets (pixels) : 

- soient $a,b \in [0,1]$ ses positions relatives aux quatre sommets (pixels) voisins,
- et $I_{00},I_{10},I_{01},I_{11}$ les quatre intensités voisines (valeurs des pixels).

La valeur du pixel $(x, y)$ obtenue par interpolation bilinéaire vérifie :

$I(x,y) = (1-a)(1-b)I_{00}+a(1-b)I_{10}+(1-a)bI_{01}+abI_{11}.$

C’est une moyenne pondérée des quatre voisins : les poids sont positifs et leur somme vaut 1. 
Le résultat est généralement plus lisse que celui du plus proche voisin, au prix d’un léger flou possible.

**Indications.** 

- Comprendre cette relation sur un schéma.
- Vérifier l'interpolation aux pixels voisins. 

### Bords de l'image

Aux bords, certaines coordonnées sortent de l’image. 
Il faut définir une règle : valeur constante, répétition du bord, réflexion ou restriction aux points valides (_crop_). 

Les calculs intermédiaires sont effectués en flottant ; convertir trop tôt en entier introduit des arrondis.
