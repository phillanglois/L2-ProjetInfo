Par exemple, `t[..., 0]` est équivalent à `t[:, :, 0]`.

## Modèle discret d'une image

Une image en niveaux de gris est une fonction discrète I[y,x], définie sur une grille de H lignes et W colonnes.  
Pour une image 8 bits, chaque intensité est comprise entre 0 et 255 : de 0 pour le noir à 255 pour le blanc.
En NumPy, l'accès se fait par `I[y, x]` : ligne puis colonne.

Une image RGB possède trois valeurs par pixel. On parle aussi de trois "canaux" R, G ou B.  
Pour une image non compressée de C canaux et de 8 bits par canal, la taille théorique est H × W × C octets.  

Les images sont représentées par des tableaux numpy : niveaux de gris `(hauteur, largeur)` ou couleurs `(hauteur, largeur, 3)`.

L'histogramme compte les pixels de chaque intensité, mais ne conserve pas leur position : des images différentes peuvent avoir le même histogramme.
