## Rappel — Modèle discret d'une image
Une image en niveaux de gris est une fonction discrète I[y,x], définie sur une grille de H lignes et W colonnes. Pour une image 8 bits, chaque intensité est comprise entre 0 et 255. En NumPy, l'accès se fait par `I[y, x]` : ligne puis colonne.

Une image RGB possède trois valeurs par pixel. Pour une image non compressée de C canaux et de 8 bits par canal, la taille théorique est H × W × C octets. L'histogramme compte les pixels de chaque intensité, mais ne conserve pas leur position : des images différentes peuvent avoir le même histogramme.

Les images sont représentées par des tableaux numpy : niveaux de gris `(hauteur, largeur)` ou couleurs `(hauteur, largeur, 3)`.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path

def mse(a, b):
    return float(np.mean((np.asarray(a, dtype=float) - np.asarray(b, dtype=float))**2))

def psnr(a, b, maxval=255):
    err = mse(a, b)
    return np.inf if err == 0 else 10*np.log10(maxval**2/err)
```

## Exercice 2 — RGB vers niveaux de gris

**Énoncé.** Écrire une fonction qui transforme une image RGB en niveaux de gris avec la formule approchée `Y = 0.299 R + 0.587 G + 0.114 B`. Retourner un tableau flottant.

## Correction 2

La conversion pondérée reflète mieux la sensibilité visuelle que la moyenne simple des trois canaux.

```python
def rgb_vers_gris(rgb):
    rgb = np.asarray(rgb, dtype=float)
    if rgb.ndim != 3 or rgb.shape[-1] < 3:
        raise ValueError("Attendu : tableau (H, W, 3) ou plus")
    return 0.299*rgb[..., 0] + 0.587*rgb[..., 1] + 0.114*rgb[..., 2]

exemple_rgb = np.array([[[255, 0, 0], [0, 255, 0], [0, 0, 255]]], dtype=np.uint8)
print(rgb_vers_gris(exemple_rgb))
```
