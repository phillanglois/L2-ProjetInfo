## Transformations affines

Une transformation affine du plan s'écrit $p' = A p + t$, où $p'$ est l'image du point $p$. $t$ est un vecteur translation et $A$ est l'opérateur matriciel d'une transformation euclidienne (translation, rotation, symétrie), d'une homothétie (changement d'échelle), d'un cisaillement (étirement selon un axe) ... 

En coordonnées homogènes, on utilise un vecteur $(x,y,1)^t$ et la transformation affine est maintenant décrite par une matrice 3×3.  

Translation, homothétie et rotation sont ainsi représentées par des matrices 3×3.

Ces transformations ne commutent généralement pas. Avec des vecteurs colonnes, M₂M₁ applique d'abord M₁, puis M₂. 

## Conventions

On utilise des coordonnées homogènes `(x, y, 1)`. 

En affichage d’image, l’axe vertical des tableaux croît vers le bas ; cela peut inverser l’intuition habituelle sur les rotations.

L'opérateur `@` est le produit matriciel (non commutatif) _numpy_

```python
# 3 transformations géométriques du plan à coordonnées homogènes
def translation(tx, ty):
    return np.array([[1, 0, tx], [0, 1, ty], [0, 0, 1]], dtype=float)

def homothetie(s):
    return np.array([[s, 0, 0], [0, s, 0], [0, 0, 1]], dtype=float)

def rotation(theta):
    c, s = np.cos(theta), np.sin(theta)
    return np.array([[c, -s, 0], [s, c, 0], [0, 0, 1]], dtype=float)
```

## Application aux images

Pour construire une image transformée, le **mappage inverse** parcourt chaque pixel de sortie et cherche sa position dans l’image d’entrée. 

Le mappage direct peut laisser des trous : les coordonnées transformées ne tombent pas toujours sur des pixels entiers. 

Attention aussi à l’axe vertical des images, dirigé vers le bas.
