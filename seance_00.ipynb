### Tests unitaires et nombres flottants

- Les nombres flottants, qui approchent les réels sur un ordinateurs, sont des nombres _à précision finie_. 
Ainsi $\pi$, bien sûr mais aussi $1/10$, ne sont pas exactement réprésentés avec un nombre fini de bits en base 2.
En revanche, $1/2, 1/4, \dots$ le sont. 
Cette précision finie est à l'origine des _erreurs d'arrondis_ qui entâchent la représentation et les calculs en nombres flottants.      
- Il n'est pas nécessaire de faire un nombre important de calculs pour observer une erreur très importante.
L'exemple suivant illustre ce danger (un résultat sans aucun chiffre significatif après deux "additions" seulement !) et aussi, plus qualitativement, la _perte de l'associativité de l'addition de nombres flottants_. 

```python
a = 10**18
un = 1.
un_a_priori = (un + a) - a
un_bien_sur =  un + (a - a) 
print(un_a_priori, un_bien_sur)
```

- L'écriture d'assertion avec égalité `==` entre expression et valeurs flottantes sont donc fausses en général.

- python propose la fonction `math.isclose()` qui permet d'écrire des assertions avec prise en compte d'une incertitude relative ou absolue.

```python
import math

a = 0.1 + 0.2
b = 0.3

# Mauvaise pratique à cause des arrondis
# assert a == b 

# Bonne pratique
assert math.isclose(a, b, rel_tol=1e-9, abs_tol=0.0)
```
